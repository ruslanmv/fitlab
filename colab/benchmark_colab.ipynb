````python
#@title 3 · Your result + one-click submission link
import glob, json, urllib.parse
REPO = "OWNER/llm-fitlab"
result = json.loads(open(sorted(glob.glob("results/*.json"))[-1]).read())
print(json.dumps(result["metrics"], indent=2))
body = "```json\n" + json.dumps(result, indent=2) + "\n```"
title = f"[bench] {result['ollama_tag']} on {result['gpu']['name']}"
url = (f"https://github.com/{REPO}/issues/new?"
       + urllib.parse.urlencode({"labels": "benchmark-submission", "title": title, "body": body}))
print("\n👉 Submit your result (opens a prefilled issue):\n" + url)
````
